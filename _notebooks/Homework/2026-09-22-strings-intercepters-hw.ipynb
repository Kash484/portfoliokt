## Popcorn Hack 1: Extract the Hidden Message

**Task:** An intercepted transmission has a real message hidden inside decoy characters at fixed positions. Use only **indexing and slicing** — no string methods — to pull the real words out.

```python
# CODE_RUNNER: Popcorn Hack 1 - Extract the hidden message using indexing and slicing

transmission = "xxSIGNALxxxCONFIRMEDxx77xxDELTA"

word_1 = transmission[2:8]
print("Word 1:", word_1)

word_2 = transmission[11:20]
print("Word 2:", word_2)

word_3 = transmission[-5:]
print("Word 3:", word_3)

print("Total length:", len(transmission))

print("Every other character:", transmission[::2])
```

## Popcorn Hack 2: Scrub and Search a Noisy Transmission

**Task:** This transmission is full of noise — stray punctuation, extra whitespace, and mismatched case. Clean it up and search it using only **string methods**.

```python
# CODE_RUNNER: Popcorn Hack 2 - Clean and search a noisy transmission using string methods

noisy = "   ...AGENT>>>Falcon---reports: the PACKAGE has ARRIVED!!!   "

trimmed = noisy.strip()
print("Trimmed:", repr(trimmed))

cleaned = trimmed.replace(">", "").replace("-", "").replace("!", "")
print("Cleaned:", cleaned)

lowered = cleaned.lower()
print("Lowered:", lowered)

has_arrived = "arrived" in lowered
print("Contains 'arrived':", has_arrived)

words = cleaned.split()
print("Word count:", len(words))
```

## Popcorn Hack 3: Build a Decoded Report

**Task:** Combine fragments recovered from the last two cycles into a single, formatted intercept report using **concatenation, `str()`, f-strings, and `.join()`**.

```python
# CODE_RUNNER: Popcorn Hack 3 - Build a decoded report using concatenation and formatting

agent_name = "Falcon"
target_word = "MIDNIGHT"
confidence = 87          # a percentage, stored as an int
fragments = ["ALPHA", "SEVEN", "DELTA"]

headline = "Agent " + agent_name + " intercepted: " + target_word
print(headline)

confidence_line = "Confidence: " + str(confidence) + "%"
print(confidence_line)

confidence_line_fstring = f"Confidence: {confidence}%"
print(confidence_line_fstring)

joined_fragments = "-".join(fragments)
print("Joined fragments:", joined_fragments)

final_report = f"REPORT: {headline} | {confidence_line} | Fragments: {joined_fragments}"
print(final_report)
```

## Homework Hack: Crack the Final Transmission

**Task:** One transmission is left, and it needs **all three skills** — indexing/slicing, string methods, and formatting — to fully decode. The transmission arrives as a **comma-separated (CSV-style) row**. The finished program must:

1. **Clean and split it.** Use `.strip()` to remove outer whitespace, then `.split(",")` to break the CSV row into fields.
2. **Extract by index.** Use indexing on the resulting list to pull out the date, the agent codename, the raw message code, and the confidence score.
3. **Slice-check the date.** Use slicing (not a method) to confirm the date field starts with the year `"2026"`.
4. **Clean and search the message.** Use `.replace()` to turn the dashes in the message code into spaces, then use `.lower()` and the `in` operator to check whether the word `"dawn"` appears anywhere in it.
5. **Rebuild it as a Markdown table row.** Use an f-string with `|` separators.
6. **Rebuild it as a JSON line.** Use an f-string to build a JSON-style line, using lowercase `true`/`false` for the boolean.

```python
# CODE_RUNNER: Homework - Crack the Final Transmission using slicing, methods, and formatting

raw_transmission = "  2026-09-15,GHOST,THE-EAGLE-LANDS-AT-DAWN,87  "

segments = raw_transmission.strip().split(",")
print("Segments:", segments)

date = segments[0]
agent = segments[1]
message_code = segments[2]
confidence_str = segments[3]
print("Date:", date, "| Agent:", agent, "| Message code:", message_code, "| Confidence:", confidence_str)

year_check = date[:4]
print("Year check:", year_check)

message = message_code.replace("-", " ")

found_dawn = "dawn" in message.lower()
print("Found 'dawn':", found_dawn)

markdown_row = f"| {date} | {agent} | {message} | {confidence_str}% |"
print("Markdown row:", markdown_row)

#         Braces are built by concatenation (not f-string brace-escaping) so the
#         raw source never contains a literal double-brace for the site builder to trip on.
json_body = (
    f'"date": "{date}", "agent": "{agent}", "message": "{message}", '
    f'"confidence": {confidence_str}, "contains_dawn": {str(found_dawn).lower()}'
)
json_line = "{" + json_body + "}"
print("JSON line:", json_line)
```

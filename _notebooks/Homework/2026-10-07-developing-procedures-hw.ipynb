## Part 1: the Dessert Robot in Python

The lesson's version wraps the call in `if __name__ == '__main__':`. That line only means "run this when the file is run directly," so here the call is made directly.


```python
# CODE_RUNNER: 3.13 Part 1 - the Dessert Robot

def make_dessert():
    measure_flour()
    mix_ingredients()
    bake()

def measure_flour():
    print("Measuring 2 cups of flour.")

def mix_ingredients():
    print("Mixing flour, sugar, eggs.")

def bake():
    print("Baking at 350°F for 25 minutes.")

# Run the procedure
make_dessert()
```

## Part 2: the mini-game, without `input()`

`input()` is unavailable in the notebook runner, so `play_game` takes the procedure name, ingredient, and run choice as parameters.

```python
# CODE_RUNNER: 3.13 Part 2 - name-the-procedure mini-game

def mix(item):
    print(f"Mixing {item} into the bowl.")

def heat():
    print("Heating the oven to a friendly temperature...")

def finish():
    print("Dessert is ready!")

def play_game(name, ingredient, run_now):
    print("Mini-game: name a procedure that chooses an ingredient, mixes it, heats, and finishes the dessert")
    if name == "":
        print("No name entered, using default: prepare_ingredient")
        name = "prepare_ingredient"

    def dynamic_procedure():
        mix(ingredient)
        heat()
        finish()

    # Store the new procedure under the chosen name
    procedures = {name: dynamic_procedure}
    print(f"Created a procedure called {name}() that does multiple steps.")
    print("Good name (readable):")
    print(f"   {name}()  # says what it does without opening it")
    print("Poor name (confusing):")
    print("   do_step()  # what does this do? you must open the function to find out")

    if run_now:
        print("Running the procedure...")
        procedures[name]()
    else:
        print("Canceled. You can run the procedure later by calling it by name.")

play_game("prepare_and_bake_dessert", "milk", True)
```

Calling `play_game("", "sugar", False)` takes the other two paths: it falls back to `prepare_ingredient` and prints the "Canceled" line instead of running.


## Part 3: the poor-naming demo

The lesson's first bad example:

    def do_step():
        print('Choosing ingredient...')
        print('Mixing...')
        print('Heating...')
        print('Finished!')

    print('Calling do_step() (confusing name):')
    do_step()

**The fix:** one procedure per step, each named for what it does, and one top-level procedure whose body reads like the recipe.


```python
# CODE_RUNNER: 3.13 Part 3 - same steps, good names

def choose_ingredient():
    print("Choosing ingredient...")

def mix_ingredient():
    print("Mixing...")

def heat_oven():
    print("Heating...")

def finish_dessert():
    print("Finished!")

def make_dessert_step_by_step():
    choose_ingredient()
    mix_ingredient()
    heat_oven()
    finish_dessert()

print("Calling make_dessert_step_by_step() (clear name):")
make_dessert_step_by_step()
```

The four step lines are **identical** to what `do_step()` prints. Good names do not change what the program does; they change how fast a person can understand it. Now if the "Heating..." line is wrong, the bug is obviously in `heat_oven()`.


## Exercise 1: rename the Python example's functions

The lesson's names are already better than `do_it()`, but they still hide details: `bake()` does not say how hot or how long, and `make_dessert()` does not say which dessert. A good name says **what** the procedure does, specifically enough that the caller never has to open it.

| Original | Renamed | Why |
|---|---|---|
| `make_dessert()` | `make_basic_cake()` | The steps (flour, sugar, eggs, bake 25 min) make a cake, not any dessert |
| `measure_flour()` | `measure_two_cups_flour()` | Says the amount, which is what the print reports |
| `mix_ingredients()` | `mix_flour_sugar_eggs()` | Says which ingredients |
| `bake()` | `bake_at_350_for_25_minutes()` | Says the temperature and time |

```python
# CODE_RUNNER: 3.13 Exercise 1 - descriptive names

def make_basic_cake():
    measure_two_cups_flour()
    mix_flour_sugar_eggs()
    bake_at_350_for_25_minutes()

def measure_two_cups_flour():
    print("Measuring 2 cups of flour.")

def mix_flour_sugar_eggs():
    print("Mixing flour, sugar, eggs.")

def bake_at_350_for_25_minutes():
    print("Baking at 350°F for 25 minutes.")

make_basic_cake()
```

**Trade-off:** very specific names break if the details change. If the recipe switched to 30 minutes, `bake_at_350_for_25_minutes()` would be lying. The better long-term fix is to make the details **parameters**, so the name describes the task and the inputs describe the numbers:


```python
# CODE_RUNNER: 3.13 Exercise 1 - details as parameters

def bake(temperature_f, minutes):
    print(f"Baking at {temperature_f}°F for {minutes} minutes.")

bake(350, 25)
bake(375, 12)
```

One procedure now covers a cake and a batch of cookies. That is the reuse that procedures are for.


## Exercise 2: the Python functions as Java methods

In Java every method lives inside a class, and every method declares a **return type** (`void` means it returns nothing) and an **access modifier**.

**What changed from Python:**

- `def make_dessert():` becomes `public static void makeDessert() { ... }`. Braces mark the body instead of indentation, every statement ends in `;`, and Java uses camelCase where Python uses snake_case.
- `print(...)` becomes `System.out.println(...)`.
- `static` means the method belongs to the class itself, so `main` can call it without creating a `DessertRobot` object first.
- The run line becomes a `main` method, which is where Java starts.

**Access: `public` vs `private`.**

- `public` methods can be called from **any** class. `makeDessert()` is public because it is the one thing other code should ask for.
- `private` methods can only be called from **inside** `DessertRobot`. The three steps are private because they are implementation details. If another class wrote `DessertRobot.bake();`, it would not compile. That protects the order: nobody outside can bake before the flour is measured.
- The benefit is that I can later rename, split, or rewrite the private steps without breaking any other class, as long as `makeDessert()` still works.

Python has no `private` keyword. The convention is a leading underscore, which tells other programmers "internal, do not call from outside." The cell below is the Python version of the same class structure, so its output can be checked here.


```python
# CODE_RUNNER: 3.13 Exercise 2 - same structure as the Java class

class DessertRobot:
    @staticmethod
    def make_dessert():          # public: the one method other code should call
        DessertRobot._measure_flour()
        DessertRobot._mix_ingredients()
        DessertRobot._bake()

    @staticmethod
    def _measure_flour():        # leading _ = private by convention
        print("Measuring 2 cups of flour.")

    @staticmethod
    def _mix_ingredients():
        print("Mixing flour, sugar, eggs.")

    @staticmethod
    def _bake():
        print("Baking at 350°F for 25 minutes.")

DessertRobot.make_dessert()
```

**Difference worth knowing:** Java **enforces** `private` (calling a private method from another class is a compile error). Python only **suggests** it: `DessertRobot._bake()` would still run from outside. The underscore is a promise between programmers, not a rule the language checks.


## Exercise 3: split a long procedure, then test the pieces

Here is one long procedure for a bake-sale order. It does four jobs at once: add up the items, apply a discount, add tax, and print a receipt.

    def process_order(items):
        subtotal = 0
        for name, price, qty in items:
            subtotal = subtotal + price * qty
        if subtotal >= 20:
            subtotal = subtotal * 0.9
        total = round(subtotal * 1.0775, 2)
        print(f"Items: {len(items)}")
        print(f"Total: ${total:.2f}")

**The split:** one procedure per job. The three math procedures **return** a number, so each can be checked with an `assert` on its own. Only the last one prints.

| Procedure | Input | Output |
|---|---|---|
| `calculate_subtotal(items)` | list of (name, price, qty) | sum of price × qty |
| `apply_bulk_discount(amount)` | a dollar amount | 10% off if it is $20 or more |
| `add_sales_tax(amount)` | a dollar amount | amount × 1.0775, rounded to cents |
| `print_receipt(items)` | list of items | prints the item count and total, using the three above |

```python
# CODE_RUNNER: 3.13 Exercise 3 - the long procedure split into four

def calculate_subtotal(items):
    subtotal = 0
    for name, price, qty in items:
        subtotal = subtotal + price * qty
    return subtotal

def apply_bulk_discount(amount):
    if amount >= 20:
        return amount * 0.9
    return amount

def add_sales_tax(amount):
    return round(amount * 1.0775, 2)

def print_receipt(items):
    total = add_sales_tax(apply_bulk_discount(calculate_subtotal(items)))
    print(f"Items: {len(items)}")
    print(f"Total: ${total:.2f}")

order = [("cupcake", 2.50, 3), ("cookie", 1.25, 2), ("cake", 12.00, 1)]
print_receipt(order)
```

### Tests: each piece on its own

Because three of the procedures return values, each one gets its own small tests. The discount tests check both sides of the \$20 line, including exactly \$20, which is where an off-by-one mistake (`>` instead of `>=`) would hide. If a test fails, its message names the exact procedure that is broken.


```python
# CODE_RUNNER: 3.13 Exercise 3 - unit tests for each procedure

def calculate_subtotal(items):
    subtotal = 0
    for name, price, qty in items:
        subtotal = subtotal + price * qty
    return subtotal

def apply_bulk_discount(amount):
    if amount >= 20:
        return amount * 0.9
    return amount

def add_sales_tax(amount):
    return round(amount * 1.0775, 2)

def print_receipt(items):
    total = add_sales_tax(apply_bulk_discount(calculate_subtotal(items)))
    print(f"Items: {len(items)}")
    print(f"Total: ${total:.2f}")

# calculate_subtotal
assert calculate_subtotal([]) == 0, "calculate_subtotal: empty order"
assert calculate_subtotal([("cookie", 1.25, 4)]) == 5.00, "calculate_subtotal: one item"
assert calculate_subtotal([("cupcake", 2.50, 3), ("cookie", 1.25, 2), ("cake", 12.00, 1)]) == 22.00, "calculate_subtotal: sample order"
print("calculate_subtotal: 3 tests passed")

# apply_bulk_discount
assert apply_bulk_discount(19.99) == 19.99, "apply_bulk_discount: just under $20"
assert apply_bulk_discount(20) == 18.0, "apply_bulk_discount: exactly $20"
assert round(apply_bulk_discount(22.00), 2) == 19.80, "apply_bulk_discount: sample order"
print("apply_bulk_discount: 3 tests passed")

# add_sales_tax
assert add_sales_tax(0) == 0, "add_sales_tax: zero"
assert add_sales_tax(100) == 107.75, "add_sales_tax: $100"
assert add_sales_tax(19.80) == 21.33, "add_sales_tax: sample order"
print("add_sales_tax: 3 tests passed")

print("All 9 tests passed")
```

- **Each test checks one job.** `apply_bulk_discount(20)` tests the discount rule alone, without needing a whole order or reading printed text.
- **Edge cases are cheap.** Testing exactly \$20 takes one line. With the long version I would have had to build an order that adds up to exactly \$20 and then work backward from the printed total.
- **Failures point to the bug.** A failing assert says `apply_bulk_discount: exactly $20`, not just "the total is wrong."
- **Reuse.** `add_sales_tax` can now be used anywhere else in the program that needs tax, not only on bake-sale orders.


## Summary

- A procedure is a named block of code that does **one** task, with parameters for its inputs.
- A good name says what the procedure does, so a reader never has to open it to find out (`make_dessert_step_by_step()` vs `do_step()`).
- Details that might change belong in **parameters**, not the name (`bake(350, 25)`).
- Java methods declare an access modifier: `public` for what other classes should call, `private` for internal steps. Python uses a leading `_` as the same signal.
- Small procedures that **return** values can each be tested on their own, and a failing test points straight to the broken piece.

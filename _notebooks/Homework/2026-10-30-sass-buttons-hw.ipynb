## Button Grammar: Overview

**Aaditya Prem (rex0dex)**

This page builds each button from the OCS button classes—without custom CSS, inline styles, or hex colors. Use a real `<button>` for an action and an `<a>` for navigation.

| Class | Purpose |
| --- | --- |
| `ocs__btn` | Provides the shared button base |
| `pill`, `fill`, `outline` | Adjust the shape and visual treatment |
| `accent`, `alert-green`, `alert-yellow`, `alert-red` | Communicate tone or urgency |
| `ocs__links` | Groups buttons and manages their spacing |

Combining these classes lets the shared `_sass` styles control each button's appearance.

## Popcorn Hack 1: Add the Base Button

The starter link had no class, so it looked like ordinary text. Adding `ocs__btn` applies the standard OCS button styling without custom CSS. I also placed the link in `ocs__links` to keep its spacing consistent with the other button groups.

```python
%%html
<!-- UI_RUNNER: Popcorn Hack 1 - Base Button Class | show_code: true -->
<div class="ocs__links">
  <a class="ocs__btn" href="https://www.mygoodbrain.org/resources">Mental Health Resource Directory</a>
</div>
```

## Popcorn Hack 2: Shape and Group Actions

- Added `pill` and `fill` to **Complete Daily Check-In**, making it a rounded, filled accent button: `ocs__btn pill accent fill`.
- Replaced the unstyled wrapper with `ocs__links` so both actions sit together with the shared spacing.

```python
%%html
<!-- UI_RUNNER: Popcorn Hack 2 - Modifiers and Button Groups | show_code: true -->
<div class="ocs__links">
  <button class="ocs__btn pill accent fill" type="button">Complete Daily Check-In</button>
  <a class="ocs__btn outline" href="https://www.mygoodbrain.org/resources">Save Draft Journal</a>
</div>
```

## Popcorn Hack 3: Set Tone and Emphasis

1. Changed the check-in button from `alert-yellow` to `alert-green` to match the positive action.
2. Applied the wide-group modifier: `ocs__links ocs__links--wide`. It keeps the standard group behavior while increasing the gap.
3. Added `fill` to **Get Crisis Support**, giving `ocs__btn alert-red fill` the strongest visual emphasis.

```python
%%html
<!-- UI_RUNNER: Popcorn Hack 3 - Status Tones and Wide Layout Bar | show_code: true -->
<div class="ocs__links ocs__links--wide">
  <button class="ocs__btn alert-green outline" type="button">Complete Check-In</button>
  <a class="ocs__btn alert-red fill" href="https://www.mygoodbrain.org/resources">Get Crisis Support</a>
</div>
```

## Homework Hack 1: Resource Toolbar

The original resource badges used inline colors, padding, and rounded corners. I removed those styles, corrected the `href` values, and chose classes that communicate each resource's priority.

| Resource | Classes | Reasoning |
| --- | --- | --- |
| Urgent Support | `ocs__btn alert-red fill pill` | Red and filled to signal urgency |
| Self-Care | `ocs__btn alert-green fill pill` | Green to suggest a positive habit |
| Community | `ocs__btn accent outline pill` | A calmer outlined treatment for an optional resource |

The three links share one `ocs__links` group. The `pill` class replaces the original rounded-corner styling.

```python
%%html
<!-- UI_RUNNER: Homework Hack 1 - Adaptive Resource Toolbar | show_code: true -->
<div class="ocs__links">
  <a class="ocs__btn alert-red fill pill" href="https://www.mygoodbrain.org/resources">Urgent Support</a>
  <a class="ocs__btn alert-green fill pill" href="https://www.mygoodbrain.org/resources">Self-Care</a>
  <a class="ocs__btn accent outline pill" href="https://www.mygoodbrain.org/resources">Community</a>
</div>
```

## Homework Hack 2: Establish Action Priority

The starter panel depended on a hardcoded card and teal/red colors. I replaced those with button classes that show the intended hierarchy.

| Priority | Action | Classes |
| --- | --- | --- |
| Primary | Log Mood Entry | `ocs__btn accent fill` |
| Secondary | View Weekly Trends | `ocs__btn outline` |
| Alert | Crisis Hotline | `ocs__btn alert-red outline` |

**Log Mood Entry** remains a `<button>` because it performs an action on this page; the other choices are `<a>` links to resources. The hotline label matches the assignment requirements. Only the primary choice is filled, while the red tone keeps the hotline easy to spot.

```python
%%html
<!-- UI_RUNNER: Homework Hack 2 - Primary vs Secondary Action Priority | show_code: true -->
<div class="ocs__links">
  <button class="ocs__btn accent fill" type="button">Log Mood Entry</button>
  <a class="ocs__btn outline" href="https://www.mygoodbrain.org/resources">View Weekly Trends</a>
  <a class="ocs__btn alert-red outline" href="https://www.mygoodbrain.org/resources">Crisis Hotline</a>
</div>
```

## What I Learned

- OCS classes can be understood as a procedure with parameters (AP CSP AAP-3.A): `ocs__btn` is the base, and modifiers such as `fill` or `alert-red` adjust its result without rewriting CSS.
- Choosing a tone is a form of selection (AAP-2.C): urgent actions use `alert-red`, positive actions use `alert-green`, and other actions can use `accent` or the default treatment.
- The base `ocs__btn` already provides a border. `outline` communicates secondary intent, while `fill` creates stronger emphasis.
- `ocs__links` controls grouping, spacing, and wrapping, so a custom flex wrapper is unnecessary.

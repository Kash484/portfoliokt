## Container Grammar: Overview

**Theme: Poway Scripps Rotary Parade**

Both exercises replace a broken starter layout with OCS containers and components. No custom CSS is needed.

### Page structure

1. `ocs__container` forms the centered outer boundary (up to 900px wide).
2. `ocs__card` groups the page content.
3. `ocs__grid ocs__grid--standard cols-3` arranges `ocs__grid-cell` items.
4. `ocs__table-wrap` and `ocs__table` present the schedule, and `ocs__callout` highlights a takeaway.

The page avoids inline styles and made-up layout classes such as `page-wrap`, `box`, `wrap`, and `tile`.

## Popcorn Hack: Rebuild the Parade Invite

The starter layout relied on a `page-wrap`, an inline-colored heading, and a flex row of three `box` elements. That row could not wrap on smaller screens.

I rebuilt it with the OCS container structure:

1. Replace `page-wrap` with one `ocs__container`.
2. Group the invite in an `ocs__card`, using `ocs__section-title` and `ocs__description` for its text.
3. Display the three details in an `ocs__grid ocs__grid--standard`, with an `ocs__grid-cell` for each one. The location uses `--accent` to highlight the key detail.
4. Remove the inline styles and the custom `box` class.

```python
%%html
<!-- UI_RUNNER: Containers Popcorn | show_code: true -->
<div class="ocs__container">
  <div class="ocs__card">
    <h2 class="ocs__section-title">Poway Rotary Parade Invite</h2>
    <p class="ocs__description">You're invited to watch the Poway Scripps Rotary Parade roll down Poway Road this Saturday.</p>
    <div class="ocs__grid ocs__grid--standard cols-3">
      <div class="ocs__grid-cell ocs__grid-cell--accent">Location: Poway Road</div>
      <div class="ocs__grid-cell">Start time: 9:00AM</div>
      <div class="ocs__grid-cell">What you get: an invaluable local and cultural experience!</div>
    </div>
  </div>
</div>
```

**Responsive check:** On a wide screen, the three cells appear in one row. Below 768px, the standard grid switches to two columns: Location and Start time share the first row, and “What you get” moves to the next. The content remains visible without squeezing or overflowing, and the container stays centered.

## Homework Hack: Build a Parade Hub

The starter used a `wrap` element, an inline-styled heading, three `tile` elements in a flex row, and a plain table with only an “entry / open” row. The rebuild uses one `ocs__container` and one `ocs__card` throughout.

1. **Heading and description:** Use `ocs__section-title` and `ocs__description`.
2. **Grid:** Use `ocs__grid ocs__grid--standard cols-3`. The `--header` cell labels the grid across its full width; Location uses `--accent`, and the rain plan uses `--muted`. Six regular cells fill the remaining space evenly: two rows of three on a wide screen or three rows of two on a phone.
3. **Schedule:** Wrap `ocs__table` in `ocs__table-wrap`. The table has an Event / Status header and two body rows, including a complete “Float entries / Open” entry.
4. **Takeaway:** Finish with a single `ocs__callout` line.

```python
%%html
<!-- UI_RUNNER: Containers Homework | show_code: true -->
<div class="ocs__container">
  <div class="ocs__card">
    <h2 class="ocs__section-title">Poway Rotary Parade</h2>
    <p class="ocs__description">Everything you need for parade day on Poway Road, all in one place.</p>
    <div class="ocs__grid ocs__grid--standard cols-3">
      <div class="ocs__grid-cell ocs__grid-cell--header">Parade Day at a Glance</div>
      <div class="ocs__grid-cell ocs__grid-cell--accent">Location: Poway Road</div>
      <div class="ocs__grid-cell">Start time: 9:00AM Saturday</div>
      <div class="ocs__grid-cell">Parade floats: local clubs and schools ride the route</div>
      <div class="ocs__grid-cell">Volunteers: greeters, setup crew, and cleanup team</div>
      <div class="ocs__grid-cell">Food drive: bring canned food to the Rotary booth</div>
      <div class="ocs__grid-cell ocs__grid-cell--muted">Rain plan: check this page the morning of the parade</div>
    </div>
    <div class="ocs__table-wrap">
      <table class="ocs__table">
        <thead>
          <tr><th>Event</th><th>Status</th></tr>
        </thead>
        <tbody>
          <tr><td>Float entries</td><td>Open</td></tr>
          <tr><td>Food drive</td><td>Collecting donations</td></tr>
        </tbody>
      </table>
    </div>
    <div class="ocs__callout">One container holds the whole Parade hub, and it looks the same on phones and computers with zero custom CSS.</div>
  </div>
</div>
```

### Responsive and Markup Checks

- **Wide screens:** The container remains centered and stops at 900px. The header spans all three grid columns, followed by two rows of three cells. The table and callout fill the card width.
- **Narrow screens (under 768px):** The grid changes to two columns. The header still spans the row, and the six cells form three balanced rows. If the table is too wide, `ocs__table-wrap` allows horizontal scrolling rather than page overflow.

**Markup check:** The page has one `ocs__container` and one `ocs__card`, no inline or embedded styles, and uses OCS classes plus the grid's `cols-3` modifier.

## What I Learned

The container provides the page boundary, the card groups related content, the grid arranges it, and cells hold the details. This nesting makes the structure easy to follow without custom CSS. The standard grid also adapts from three columns to two on smaller screens automatically, making it more reliable than the starter's fixed flex row.

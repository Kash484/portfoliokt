## Popcorn Hack

The starter used custom CSS and generic `div`/`span` elements to imitate text headings. I replaced those with semantic heading and paragraph elements, then used OCS classes for the visual styling.

```python
%%html
<!-- UI_RUNNER: Typography Popcorn Hack -->
<h2 class="ocs_section-title">Why Cats Purr</h2>
<h3>A Sign of Happiness</h3>
<p class="ocs_text">Cats usually purr when they feel safe, cozy, and ready for a nap.</p>
```

## Homework Hack

I removed the embedded styles and custom classes, then organized the content in an OCS card. An `h2` introduces the page and `h3` elements label its sections. The numbered items now use one ordered list with three list items, and `<strong>` marks the important phrase semantically instead of relying on a styled `span`.

```python
%%html
<!-- UI_RUNNER: Typography Homework -->
<div class="ocs_card">
  <h2 class="ocs_section-title">How to Pat a Cat</h2>
  <h3>Have you ever wanted to...</h3>
  <p class="ocs_text">pat a cat? Well good for you, we will give step by step instructions on how to pat a cat!</p>
  <p class="ocs_text"><strong>How to pat a cat:</strong></p>
  <ol>
    <li>Put out your hand</li>
    <li>Touch the cat</li>
    <li>Move your hand back and forth</li>
  </ol>
  <h3>Congrats! You pat a cat :D</h3>
</div>
```

## Takeaway

Semantic HTML explains the role of each piece of content to browsers and assistive technology. OCS classes provide the site's visual style, so headings do not need hardcoded font sizes.
